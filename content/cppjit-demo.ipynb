```c++
class MyClass {
public:
    MyClass(int i) : m_data(i) {}
    virtual ~MyClass() {}
    virtual int add_int(int i) { return m_data + i; }
    int m_data;
};
```

```c++
int callback(MyClass* m, int i) {
    return m->add_int(i);
}
```

```c++
%%python

MyClass = gbl.MyClass
m = MyClass(0)
m
```

```c++
%%python

class PyMyClass(MyClass):
    def add_int(self, i):  # python side override (CPython only)
        return self.m_data + 2*i
```

```c++
%%python
gbl.callback(m, 2)
```

```c++
%%python
gbl.callback(PyMyClass(1), 2)
```
